1.  What is the difference between a function and a method in Python?
   - In Python, a function is a block of reusable code that performs a specific task and is defined using the `def` keyword, usually at the top level of a module or script. A method, on the other hand, is a function that is associated with an object and is called using dot notation on that object—it is defined inside a class and operates on instances of that class. For example, `len([1, 2, 3])` calls a built-in function `len()`, whereas `[1, 2, 3].append(4)` calls the `append()` method on a list object. The key distinction is that methods automatically receive the instance (`self`) as their first argument, while functions do not.

2. Explain the concept of function arguments and parameters in Python.
   - In Python, parameters are the variable names listed in a function’s definition, while arguments are the actual values passed to the function when it is called. Parameters act as placeholders for the values that will be supplied later. For example, in the function definition `def greet(name):`, `name` is a parameter. When the function is called like `greet("Alice")`, the value `"Alice"` is the argument. Python supports different types of arguments such as positional arguments, keyword arguments, default arguments, and variable-length arguments (`*args` and `**kwargs`), allowing flexibility in how functions are called and defined.

3. What are the different ways to define and call a function in Python?
   - In Python, there are several ways to define and call functions, depending on how flexible or specialized the function behavior needs to be. The most common method is using the `def` keyword to define a standard function, such as `def add(a, b): return a + b`, which can be called using positional arguments like `add(2, 3)` or keyword arguments like `add(a=2, b=3)`. Functions can also have default argument values, allowing calls with fewer arguments than defined; for instance, `def greet(name="Guest")` can be called as `greet("Alice")` or simply `greet()` to use the default. Variable-length arguments provide further flexibility: `*args` collects extra positional arguments into a tuple, while `**kwargs` collects extra keyword arguments into a dictionary. For example, `def total(*args): return sum(args)` can be called as `total(1, 2, 3)`, and `def show_info(**kwargs): print(kwargs)` can be called as `show_info(name="John", age=25)`. Python also supports anonymous functions using `lambda`, such as `square = lambda x: x**2`, which can be called with `square(4)`. Functions can be nested, meaning one function can be defined inside another; for example, `def outer(): def inner(): return "Hello"; return inner()` allows calling `outer()` to execute the inner function. Lastly, the `functools.partial` method can be used to pre-fill some arguments of a function, creating a new callable with fewer parameters—for example, `from functools import partial`, `double = partial(lambda x, y: x * y, 2)` lets you call `double(5)` to get `10`. These different styles offer powerful ways to write concise and reusable code.

4. What is the purpose of the `return` statement in a Python function?
   - The `return` statement in a Python function is used to exit the function and send a result back to the caller. It allows the function to produce a value that can be used elsewhere in the program, enabling data flow and modular design. Without a `return` statement, a function returns `None` by default. For example, in `def add(a, b): return a + b`, the `return` statement outputs the sum of `a` and `b`, allowing a call like `result = add(2, 3)` to store the value `5` in `result`. The `return` statement can also return multiple values as a tuple, such as `return x, y`, which can be unpacked as `a, b = func()`.

5. What are iterators in Python and how do they differ from iterables?
   - In Python, an iterator is an object that represents a stream of data; it implements the `__iter__()` and `__next__()` methods, allowing it to return items one at a time when requested. An iterable, on the other hand, is any object that can be looped over (such as lists, tuples, or strings) and implements the `__iter__()` method, which returns an iterator. The key difference is that iterables are containers you can iterate over multiple times, while iterators produce items one by one and get exhausted after one pass. For example, a list like `[1, 2, 3]` is an iterable, and calling `iter([1, 2, 3])` returns an iterator. Using `next()` on the iterator will give `1`, then `2`, then `3`, and finally raise a `StopIteration` error when items are exhausted.

6. Explain the concept of generators in Python and how they are defined.
   - Generators in Python are a type of iterable that produce values lazily, meaning they yield one item at a time and only when requested, which makes them memory-efficient for large datasets. They are defined using regular functions but with the `yield` keyword instead of `return`. Each time the generator's `__next__()` method is called (e.g., using `next()`), execution resumes from where it last yielded a value. For example, a simple generator can be defined as `def count_up_to(n): i = 1; while i <= n: yield i; i += 1`. Calling this function like `gen = count_up_to(3)` creates a generator object, and using `next(gen)` repeatedly will yield `1`, then `2`, then `3`, after which it raises `StopIteration`. Generators are especially useful for reading large files, generating infinite sequences, or streaming data without storing everything in memory.

7. What are the advantages of using generators over regular functions?
   - Generators offer several advantages over regular functions, especially when dealing with large datasets or streams of data. The most significant benefit is memory efficiency —since generators yield one value at a time, they do not store the entire output in memory, unlike regular functions that return all values at once. This makes generators ideal for processing big data or infinite sequences, like reading large log files line by line using `def read_lines(file): for line in file: yield line`. Additionally, generators have lazy evaluation, meaning they produce values only when needed, which improves performance and reduces unnecessary computation. They also simplify code by eliminating the need to manage intermediate states manually; for example, instead of using a class with `__iter__()` and `__next__()` methods to implement a custom iterator, a generator function with `yield` can achieve the same result in fewer lines.

8. What is a lambda function in Python and when is it typically used?
   - A lambda function in Python is an anonymous, small, and simple function defined using the `lambda` keyword instead of the standard `def`. It can take any number of arguments but only contains a single expression, which is evaluated and returned. The syntax is `lambda arguments: expression`. Lambda functions are often used for short-term operations where a full function definition would be unnecessary. For example, `square = lambda x: x ** 2` creates a lambda function that squares a number, and it can be used in expressions like `map(lambda x: x**2, [1, 2, 3])` or `sorted(list_of_tuples, key=lambda x: x[1])` to sort by the second item of each tuple. Typically, lambda functions are used in situations where a function is required as an argument, such as in functional programming tools like `map()`, `filter()`, and `reduce()`, or when creating simple callbacks.

9. Explain the purpose and usage of the `map()` function in Python.
   -The `map()` function in Python is used to apply a specified function to each item in an iterable (like a list or tuple) and return a map object (which is an iterator). The purpose of `map()` is to simplify the process of applying a function to every element of an iterable without the need for an explicit loop. The syntax is `map(function, iterable)`, where the function is applied to each element of the iterable. For example, `map(lambda x: x**2, [1, 2, 3])` applies the lambda function that squares each element of the list, resulting in a map object that, when converted to a list, yields `[1, 4, 9]`. This is often used when you need to process data in a concise, functional manner. The `map()` function can also accept multiple iterables, in which case the function must accept as many arguments as there are iterables. For example, `map(lambda x, y: x + y, [1, 2, 3], [4, 5, 6])` adds corresponding elements from the two lists, resulting in `[5, 7, 9]`.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
   - The `map()`, `reduce()`, and `filter()` functions in Python are all functional programming tools used to process iterables, but they serve different purposes. The `map()` function applies a given function to each item in an iterable, returning an iterator with the transformed elements. It’s useful for modifying or transforming each element individually. For example, `map(lambda x: x**2, [1, 2, 3])` squares each element of the list, yielding `[1, 4, 9]`. The `reduce()` function, from the `functools` module, applies a binary function cumulatively to the items of an iterable, reducing it to a single value. This is helpful when performing a cumulative operation like summing or multiplying all elements. For instance, `reduce(lambda x, y: x + y, [1, 2, 3])` returns `6`, the sum of the elements. The `filter()` function, on the other hand, applies a function to each item in an iterable and returns an iterator containing only those elements for which the function evaluates to `True`. This is commonly used to filter out elements that don't meet a certain condition. For example, `filter(lambda x: x > 2, [1, 2, 3])` filters out numbers less than or equal to 2, yielding `[3]`.

```python
#1.  Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in the list.

def sum_of_even_numbers(numbers):
    return sum(filter(lambda x: x % 2 == 0, numbers))

# Example usage:
numbers = [1, 2, 3, 4, 5, 6]
result = sum_of_even_numbers(numbers)
print(result)
```

```python
#2.  Create a Python function that accepts a string and returns the reverse of that string.

def reverse_string(s):
    return s[::-1]

# Example usage:
input_string = "hello"
result = reverse_string(input_string)
print(result)
```

```python
#3.  Implement a Python function that takes a list of integers and returns a new list containing the squares of each number.

def square_numbers(numbers):
    return [x**2 for x in numbers]

# Example usage:
numbers = [1, 2, 3, 4, 5]
result = square_numbers(numbers)
print(result)
```

```python
#4.  Write a Python function that checks if a given number is prime or not from 1 to 200.

def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

# Example usage:
for number in range(1, 201):
    if is_prime(number):
        print(number, end=" ")
```

```python
#5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of terms.

class FibonacciIterator:
    def __init__(self, terms):
        self.terms = terms
        self.current = 0
        self.previous = 0
        self.next_term = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < self.terms:
            if self.count == 0:
                self.count += 1
                return self.current
            elif self.count == 1:
                self.count += 1
                return self.next_term
            else:
                fib = self.current + self.next_term
                self.current = self.next_term
                self.next_term = fib
                self.count += 1
                return fib
        else:
            raise StopIteration
```

```python
#6.  Write a generator function in Python that yields the powers of 2 up to a given exponent.

def powers_of_2(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage:
for power in powers_of_2(5):
    print(power, end=" ")
```

```python
#7.  Implement a generator function that reads a file line by line and yields each line as a string.

def read_file_line_by_line(file):
    for line in file:
        yield line.strip()
```

```python
#8. Use a lambda function in Python to sort a list of tuples based on the second element of each tuple.

tuples_list = [(1, 3), (4, 1), (2, 5), (3, 2)]
sorted_list = sorted(tuples_list, key=lambda x: x[1])
print(sorted_list)
```

```python
#9.  Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

celsius_temps = [0, 20, 37, 100]

def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

fahrenheit_temps = map(celsius_to_fahrenheit, celsius_temps)
print(list(fahrenheit_temps))
```

```python
#10.  Create a Python program that uses `filter()` to remove all the vowels from a given string.

def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda x: x not in vowels, s))

# Example usage
input_string = "Hello World"
result = remove_vowels(input_string)
print(result)
```

```python
#11.  Imagine an accounting routine used in a book shop. It works on a list with sublists, which look like this:
#Write a Python program, which returns a list with 2-tuples. Each tuple consists of the order number and the product of the price per item and the quantity. The product should be increased by 10,- € if the value of the order is smaller than 100,00 €.

#Write a Python program using lambda and map.

data = [
    ("34587", "Learning Python, Mark Lutz", 4, 40.95),
    ("98762", "Programming Python, Mark Lutz", 5, 56.80),
    ("77226", "Head First Python, Paul Barry", 3, 32.95),
    ("88112", "Einführung in Python3, Bernd Klein", 3, 24.99),
]

def calculate_product(item):
    order_number, _, quantity, price_per_item = item
    product = quantity * price_per_item
    if product < 100:
        product += 10
    return (order_number, product)

result = list(map(lambda item: calculate_product(item), data))

print(result)


```
